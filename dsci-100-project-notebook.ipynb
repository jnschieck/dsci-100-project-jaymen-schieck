```r
options(repr.plot.width = 8, repr.plot.height = 6)
reg_win_pct_hist <- tidy_data |> #histogram showing the distribution of regular season win percentage
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(bins = 11) +
    labs(x = "Regular Season Win Percentage (%)", y = "Number of Teams", title = "Distribution of Regular Season Win Percentage") +
    theme(text = element_text(size = 18))
reg_win_pct_hist
```

```r
options(repr.plot.width = 5, repr.plot.height = 4)
won_first_game_prop <- tidy_data |> #calculating the number of teams that won or lost their first game of the tournament
    group_by(won_first_game) |>
    summarize(proportion = n())

won_first_game_bar <- won_first_game_prop |> #bar chart of these proportions
    ggplot(aes(x = won_first_game, y = proportion)) +
    geom_bar(stat = "identity") +
    labs(x = "Won First Game of Tournament", y = "Number of Teams", title = "Proportion of Teams That Won Their First Game") +
    theme(text = element_text(size = 12))
won_first_game_bar
```

The scatterplot shows seed and regular season win percentage coloured by the outcome of the first tournament game. This shows how K-nearest neighbors will classify teams based on similar points from the predictors. It also shows that higher seeds tend to have higher win percentages, showing that the predictors are related. The higher and lower seeds almost all won and lost their first game, respectively, meaning these regions should be predicted fairly accurately. The middle seeds have a bigger overlap with winners and losers, so the predictions may be less accurate.

```r
options(repr.plot.width = 12, repr.plot.height = 10)
seed_win_pct_plot <- tidy_data |> #scatterplot of seed vs. regular season win percentage, coloured by teams winning or losing their first tournament game
    ggplot(aes(x = seed, y = reg_wins_pct, colour = won_first_game)) +
    geom_point(alpha = 0.8) +
    labs(x = "Tournament Seed", y = "Regular Season Win Percentage (%)", colour = "Won First Game", title = "Tournament Seed vs. Regular Season Win Percentage (%)") +
    theme(text = element_text(size = 18))
seed_win_pct_plot
```

The bar graph shows how the tournament seed relates to the outcome to be predicted. Higher seeds almost always won their first game, and the win rate decreases as the tournament seed increases. This tells us that the tournament seed will be a strong predictor.

```r
options(repr.plot.width = 7, repr.plot.height = 6)
seed_bar <- tidy_data |> #bar plot showing how many teams of each seed won or lost their first tournament game
    ggplot(aes(x = as_factor(seed), fill = won_first_game)) +
    geom_bar() +
    labs(x = "Tournament Seed", y = "Number of Teams", fill = "Won First Game", title = "Outcome of First Tournament Game by Seed") +
    theme(text = element_text(size = 15))
seed_bar
```
